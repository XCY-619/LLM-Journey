```python
import torch
import torch.nn as nn
import math
torch.manual_seed(42)

B, T, C = 1, 5, 128 # 为了观察清晰，我们只看 1 个批次，5 个 Token，128 维特征

# 1. 模拟输入并进行 Q, K 线性投影
x = torch.randn(B, T, C) 
Q = nn.Linear(C, C)(x)
K = nn.Linear(C, C)(x)

# 2. 原始注意力分数 (Raw Scores)
raw_scores = Q @ K.transpose(-2, -1)
print("=== 1. 原始注意力矩阵 (Raw Scores) ===")
print("形状:", raw_scores.shape)
print(raw_scores[0].detach().numpy().round(2)) # 打印 5x5 方阵，保留两位小数

# 3. 缩放注意力分数 (Scaled Scores)
d_k = C
scaled_scores = raw_scores / math.sqrt(d_k)
print("\n=== 2. 除以根号 d_k 后的矩阵 (Scaled Scores) ===")
print(scaled_scores[0].detach().numpy().round(2))

# 4. Softmax 归一化权重 (Attention Weights)
import torch.nn.functional as F
attention_weights = F.softmax(scaled_scores, dim=-1)
print("\n=== 3. Softmax 归一化后的矩阵 (Attention Weights) ===")
print(attention_weights[0].detach().numpy().round(3))
# 验证每行求和是否为 1
print("\n验证第一行的权重和:", attention_weights[0][0].sum().item())
```
